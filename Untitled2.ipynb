First, let's preprocess the data by dropping irrelevant columns and converting categorical features to numeric using one-hot encoding.

```python
df_processed = df.copy()

# Drop identifier columns that are not relevant for correlation
df_processed = df_processed.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)

# Convert categorical features to numeric using one-hot encoding
df_processed = pd.get_dummies(df_processed, columns=['Geography', 'Gender'], drop_first=True)

# Display the first few rows of the processed DataFrame
display(df_processed.head())
```

Now that the data is preprocessed with only numerical columns, we can calculate the correlation matrix and visualize it with a heatmap, focusing on the 'Exited' column.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix
corr_matrix = df_processed.corr()

# Create a heatmap of the correlation matrix
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Features including Exited')
plt.show()

# Display correlations with the target variable 'Exited'
plt.figure(figsize=(8, 6))
corr_with_exited = corr_matrix['Exited'].sort_values(ascending=False)
sns.barplot(x=corr_with_exited.values, y=corr_with_exited.index, palette='viridis')
plt.title('Correlation of Features with Exited')
plt.xlabel('Correlation Coefficient')
plt.ylabel('Features')
plt.show()
```

Before applying any resampling techniques or training models, we need to separate our features (X) from our target variable (y).

```python
# Separate features (X) and target (y)
X = df_processed.drop('Exited', axis=1)
y = df_processed['Exited']

print(f"Original target distribution:\n{y.value_counts()}")
```

### 1. Handling Imbalance with SMOTE (Oversampling)

SMOTE is a popular oversampling technique that generates synthetic samples from the minority class. We'll use `imblearn` library for this.

```python
from imblearn.over_sampling import SMOTE
from collections import Counter

print('Original dataset shape %s' % Counter(y))

sm = SMOTE(random_state=42)
X_res, y_res = sm.fit_resample(X, y)

print('Resampled dataset shape %s' % Counter(y_res))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_res,y_res,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()

X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.fit_transform(X_test)
```

```python
X_train.sample(5)
```

```python
X_train_scaled.shape
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()

# 2(experiment) Hidden layers for 11 input vectors + 1 output layer
model.add(Dense(11,activation='relu',input_dim=11)) # 11 nodes
model.add(Dense(10,activation='relu'))  # 10 nodes
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
train_logs=model.fit(X_train_scaled,y_train,epochs=100, validation_split=0.2)
# while training also validate the data on a splitted validation data(20% of X_train_scaled)
```

```python
model.layers[0].get_weights()
```

```python
y_result=model.predict(X_test_scaled)
# The outputs from the neural network aren't 0 or 1..since they are the outputs
#from sigmoid function which are probabilities of belonging to a class
```

```python
y_pred=np.where(y_result>=0.5,1,0) # where y_result is >=0.5 make it 1 else 0
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_test,y_pred))
```

```python
history=train_logs.history
```

```python
plt.plot(history['loss'])
plt.plot(history['val_loss'])
plt.title("Model Loss")
plt.ylabel("Loss")
plt.xlabel("Epoch")
```

```python
plt.plot(history['accuracy'])
plt.plot(history['val_accuracy'])
plt.title("Model Accuracy")
plt.ylabel("Accuracy")
plt.xlabel("Epoch")
```
